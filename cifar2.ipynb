```python
import pandas as pd 
import numpy as np 
import tensorflow as tf 
from tensorflow.keras import layers, models 
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import LabelEncoder 
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv(r"E:\DeepLearning\cnn_cifar task2\sonar_dataset.csv", header=None)
```

```python
print(df.head())
```

```python
X = df.drop(60, axis='columns') 
Y = df[60]
```

```python
print("Input Shape :", X.shape) 
print("Output Sample :")
print(Y[:5])
```

```python
encoder = LabelEncoder()
y = encoder.fit_transform(Y) 
print("Encoded Labels :") 
print(y[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split( 
    X, y, 
    test_size=0.2, 
    random_state=1 
    ) 
print("Training Shape :", X_train.shape) 
print("Testing Shape :", X_test.shape)
```

```python
X_train = X_train / X_train.max() 
X_test = X_test / X_test.max()
```

```python
X_train = X_train.values.reshape(-1, 60, 1) 
X_test = X_test.values.reshape(-1, 60, 1) 
print("Reshaped Training Shape :", X_train.shape)
```

```python
cnn = models.Sequential([ 
    layers.Conv1D( filters=32, 
                  kernel_size=3, 
                  activation='relu', 
                  input_shape=(60,1) 
                  ), layers.MaxPooling1D(2),  layers.Conv1D( 
                      filters=64, 
                      kernel_size=3, 
                      activation='relu' 
                      ),  
                      layers.MaxPooling1D(2),  
                      layers.Flatten(),  
                      layers.Dense(64, activation='relu'),  
                      layers.Dense(1, activation='sigmoid') ])
```

```python
cnn.compile( 
    optimizer='adam', 
            loss='binary_crossentropy',
             metrics=['accuracy'] 
             )
```

```python
cnn.fit( X_train, y_train, epochs=20 )
```

```python
cnn.evaluate(X_test, y_test)
```

```python
y_pred = cnn.predict(X_test)
```

```python
y_pred_classes = [] 
for element in y_pred: 
    if element > 0.5: 
        y_pred_classes.append(1) 
    else: y_pred_classes.append(0) 
print(classification_report(y_test, y_pred_classes))
```
